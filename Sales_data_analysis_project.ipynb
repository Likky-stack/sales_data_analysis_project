```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
import pandas as pd

df = pd.read_csv("E-commerece sales data 2024.csv")
df.head()
```

```
df.info()
```

```
df = df.drop(columns=["Unnamed: 4"])
```

```
df.isnull().sum()
```

```
df = df.dropna()
```

```
df.columns = df.columns.str.replace(" ", "_")
```

```
df['Time_stamp'] = pd.to_datetime(df['Time_stamp'], dayfirst=True)
```

```
df['Month'] = df['Time_stamp'].dt.month
```

```
df['Interaction_type'].value_counts()
```

```
df['Interaction_type'].value_counts().plot(kind='bar')

plt.title("User Interaction Types")
plt.xlabel("Interaction")
plt.ylabel("Count")

plt.show()
```

```
monthly = df.groupby('Month')['Interaction_type'].count()
monthly.plot(kind='line')

plt.title("Monthly Activity")
plt.xlabel("Month")
plt.ylabel("Interactions")

plt.show()
```
